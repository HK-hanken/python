```python
#Task 1

transactions = [
    {"id": 1, "product": "Laptop", "price": 1200, "quantity": 2, "region": "EU"},
    {"id": 2, "product": "Phone", "price": 800, "quantity": -1, "region": "US"},
    {"id": 3, "product": "Tablet", "price": 0, "quantity": 5, "region": "EU"},
    {"id": 4, "product": "Monitor", "price": 350, "quantity": 3, "region": "US"},
    {"id": 5, "product": "Laptop", "price": 1200, "quantity": 1, "region": "EU"},
    {"id": 6, "product": "Keyboard", "price": 75, "quantity": 10, "region": "APAC"},
]

# a)
valid_transactions = [
    t for t in transactions if t["price"]>0 and t["quantity"]>=0
]

# b)
for t in valid_transactions:
    t["total"]=t["price"]*t["quantity"]

# c)
revenue_by_region = {}
for t in valid_transactions:
    region = t["region"]
    revenue_by_region[region] = revenue_by_region.get(region, 0) + t["total"]

# d)
max_region = max(revenue_by_region, key=revenue_by_region.get)
print(f"Highest revenue region: {max_region} ({revenue_by_region[max_region]})")
```

```python
#Task 2

sales = [1200, 980, 1540, 870, 2010, 1650, 1300, 1100, 950, 1780, 2200, 1430, 1600, 1900]

total = sum(sales)
average = round(total/len(sales), 2)
best_day = sales.index(max(sales))+1
worst_day = sales.index(min(sales))+1
above_average_count = sum(1 for s in sales if s > average)


print("total:", total)
print("average:", average)
print("best_day:", best_day)
print("worst_day:", worst_day)
print("above_average_count:", above_average_count)
```

```python
#Task 3

students = [
    {"name": "Anna", "grade": 85, "attendance": 0.92},
    {"name": "Erik", "grade": 58, "attendance": 0.75},
    {"name": "Mia", "grade": 91, "attendance": 0.98},
    {"name": "Leo", "grade": 44, "attendance": 0.60},
    {"name": "Sara", "grade": 72, "attendance": 0.88},
    {"name": "Tom", "grade": 67, "attendance": 0.70},
    {"name": "Nina", "grade": 95, "attendance": 0.99},
    {"name": "Olli", "grade": 50, "attendance": 0.65},
]

# a)
passing = []
for s in students:
    if s["grade"] >= 50:
        passing.append(s["name"])

# b)
for s in students:
    if s["grade"] >= 80:
        print(f"{s["name"]}: Excellent")
    elif s["grade"] >= 50:
        print(f"{s["name"]}: Pass")
    else:
        print(f"{s["name"]}: Fail")

# c)
count = 0
for s in students:
    if s["attendance"] < 0.8:
        count +=1
print(count)

# d)
top_student = students[0]
for s in students:
    if s["grade"] > top_student["grade"]:
        top_student = s
print(top_student["name"],top_student["grade"])

# e)
top_student = max(students, key=lambda s: s["grade"])
print(top_student["name"],top_student["grade"])
```

```python
#Task 4

class Portfolio:
    def __init__(self, owner):
        self.owner = owner
        self.holdings = {}

    # a)
    def add_asset(self, ticker, shares, price_per_share):
        self.holdings[ticker] = {
            "shares": shares,
            "price_per_share": price_per_share
        }

    # b)
    def total_value(self):
        total = 0
        for asset in self.holdings.values():
            total += asset["shares"] * asset["price_per_share"]
        return total

    # C)
    def most_valuable(self):
        if not self.holdings:
            return None
        return max(self.holdings, key=lambda t: self.holdings[t]["shares"] * self.holdings[t]["price_per_share"])

    # d)
    def __str__(self):
        num_assets = len(self.holdings)
        return f"Portfolio Owner: {self.owner} | Total Assets: {num_assets} | Total Value: ${self.total_value():,.2f}"

my_portfolio = Portfolio("Alice")
my_portfolio.add_asset("AAPL", 10, 180.50)
my_portfolio.add_asset("TSLA", 5, 250.00)
my_portfolio.add_asset("NVDA", 20, 450.00)

print(f"Total Portfolio Value: ${my_portfolio.total_value()}")
print(f"Most Valuable Asset: {my_portfolio.most_valuable()}")
print(my_portfolio)
```

```python
#Task 5

orders = [
    ("2025-01-15", "Laptop"), ("2025-01-22", "Phone"),
    ("2025-02-03", "Laptop"), ("2025-02-14", "Tablet"),
    ("2025-03-01", "Phone"), ("2025-03-10", "Phone"),
    ("2025-03-22", "Laptop"), ("2025-04-05", "Monitor"),
]

from datetime import datetime
date_objects = [datetime.strptime(date_str, "%Y-%m-%d").date() for date_str, product in orders]
earliest_date = min(date_objects)
latest_date = max(date_objects)
delta = latest_date - earliest_date

print(f"Earliest order date: {earliest_date}")
print(f"Latest order date: {latest_date}")
print(f"Number of days between the earliest and latest order: {delta.days} days")
```

```python
#Task 6

# Rows: years (2022, 2023, 2024), Columns: quarters (Q1, Q2, Q3, Q4)

import numpy as np

revenue = np.array([
    [210, 230, 190, 250],
    [220, 260, 200, 270],
    [250, 290, 230, 310]])

# a)
annual_revenue = revenue.sum(axis=1)
print(f"Annual Revenue (2022-2024): {annual_revenue}")

# b)
mean_quarterly = revenue.mean(axis=0)
print(f"Mean Quarterly Revenue (2022-2024): {mean_quarterly}")

# c)
growth_rates = np.array([1.05, 1.03, 1.08, 1.02])
adjusted_2024 = revenue[2] * growth_rates
print(f"Adjusted 2024 Revenue : {adjusted_2024}")

# d)
exceed_250 = revenue[revenue > 250]
print(f"Quarterly Revenue Values Exceed 250: {exceed_250}")
```

```python
#Task 7

import pandas as pd

data = {
    "employee": ["Anna", "Erik", "Mia", "Leo", "Sara", "Tom"],
    "department": ["Finance", "Marketing", "Finance", "IT", "Marketing", "IT"],
    "salary": [4500, 4000, 4800, 5200, 3900, 5100],
    "years_exp": [5, 3, 7, 10, 2, 8],
    "performance_score": [4.2, 3.8, 4.5, 4.0, 3.5, 4.9]
}

df = pd.DataFrame(data)

# a)
df["salary_per_year_exp"] = (df["salary"]/df["years_exp"]).round().astype(int)

# b)
top_3_performers = df.sort_values(by="performance_score", ascending=False).head(3)
print("--- Top 3 Performers ---")
print(top_3_performers)

# c)
finance_high_performers = df.loc[(df["department"] == "Finance") & (df["performance_score"] > 4.0)]
print("\n--- Finance High Performers ---")
print(finance_high_performers)

# d)
df = df.rename(columns={"years_exp": "experience_years"})

# e)
first_four_subset = df.iloc[:4][["salary", "performance_score"]]
print("\n--- Salary and Performance (First 4 Employees) ---")
print(first_four_subset)
```

```python
#Task 8

import numpy as np
import pandas as pd

rng = np.random.default_rng(seed=123)
days = pd.date_range(start="2025-01-06", periods=28, freq="D")
visitors = rng.integers(200, 600, size=28)
conversion_rate = rng.uniform(0.02, 0.08, size=28).round(4)
avg_order_value = rng.normal(50, 10, size=28).round(2)

# a)
df = pd.DataFrame({
    "visitors": visitors,
    "conversion_rate": conversion_rate,
    "avg_order_value": avg_order_value
}, index=days)

# b)
df["conversions"] = (df["visitors"]*df["conversion_rate"]).round().astype(int)
df["daily_revenue"] = df["conversions"]*df["avg_order_value"]
print(df.head())
```
