```python
#Task 2
#a)
import requests
import pandas as pd
from dotenv import load_dotenv
import os
tickers = ["META", "AAPL", "AMZN", "NFLX", "GOOG"]
all_data = []
for ticker in tickers:
    url = "http://api.marketstack.com/v1/eod"
    params = {
    "access_key":api_key,
    "symbols":ticker,
    "date_from":"2026-03-01",
    "date_to":"2026-03-31"
    }
    response = requests.get(url, params=params)
    data = response.json()
    for item in data["data"]:
        item["ticker"] = ticker
        all_data.append(item)

#b)
df = pd.DataFrame(all_data)
display(df.head(5))

#c)
returns = {}
for ticker in tickers:
    stock = df[df["ticker"] == ticker].sort_values("date")
    start_price = stock.iloc[0]["close"]
    end_price = stock.iloc[-1]["close"]
    return_pct = (end_price - start_price) / start_price * 100
    returns[ticker] = return_pct
returns_df = pd.DataFrame(list(returns.items()), columns=["Stock", "Return%"])
display(returns_df)

#d)
best = returns_df.loc[returns_df["Return%"].idxmax()]
worst = returns_df.loc[returns_df["Return%"].idxmin()]

display("Best performer:", best)
display("Worst performer:", worst)
```

```python
#Task 8
import pandas as pd

wide_data = pd.DataFrame({ "store": ["Helsinki", "Tampere", "Turku"], "Q1_sales": [12000, 8500, 9200], "Q2_sales": [13500, 9100, 9800], "Q3_sales": [11800, 8800, 10100], "Q4_sales": [15000, 10200, 11500] })
long_data = pd.DataFrame({ "month": ["Jan", "Jan", "Feb", "Feb", "Mar", "Mar"], "product": ["Laptop", "Phone", "Laptop", "Phone", "Laptop", "Phone"], "units_sold": [45, 120, 50, 110, 55, 130] })

#a)
print("Before (Wide):") 
display(wide_data)
long_from_wide = pd.melt(
    wide_data,
    id_vars="store",
    var_name="quarter",
    value_name="sales"
)
print("After (Long):")
display(long_from_wide)
```

```python
#b)
print("Before (long):")
display(long_data)

wide_from_long = long_data.pivot(
    index="month",
    columns="product",
    values="units_sold"
)
print("After (Wide):")
display(wide_from_long)
```

```python
#c)
wide_back = long_from_wide.pivot(
    index="store",
    columns="quarter",
    values="sales"
).reset_index()
print("Wide restored:") 
display(wide_back)

long_back = wide_from_long.reset_index().melt(
    id_vars="month",
    var_name="product",
    value_name="units_sold"
)
print("Long restored:")
display(long_back)
```
